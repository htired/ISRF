```python
import pickle
import os
from sklearn.decomposition import PCA
```

```python
with open("item_feature.pkl", "rb") as f:
    llm_item_emb = pickle.load(f)
```

```python
print(llm_item_emb.shape)
```

```python
pca = PCA(n_components=512)
pca_item_emb = pca.fit_transform(llm_item_emb)
```

```python
with open(os.path.join("pca_item_feature.pkl"), "wb") as f:
    pickle.dump(pca_item_emb, f)
```
